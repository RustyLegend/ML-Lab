```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("retail.csv")

X = data[
    [
        "TotalQuantity",
        "TotalOrdering",
        "TotalSpending",
        "AverageOrderValue"
    ]
].values
```

```python
data
```

```python
mean = np.mean(X, axis=0)
std = np.std(X, axis=0)

X_scaled = (X - mean) / std
```

```python
def calculate_distances(X, centroids):

    distances = []

    for point in X:

        point_distances = []

        for centroid in centroids:

            distance = np.sqrt(
                np.sum((point - centroid) ** 2)
            )

            point_distances.append(distance)

        distances.append(point_distances)

    return np.array(distances)
```

```python
def assign_clusters(X, centroids):

    distances = calculate_distances(X, centroids)

    return np.argmin(distances, axis=1)
```

```python
def calculate_centroids(X, clusters, k):

    new_centroids = []

    for i in range(k):

        points = X[clusters == i]

        if len(points) > 0:
            centroid = np.mean(points, axis=0)
        else:
            centroid = np.zeros(X.shape[1])

        new_centroids.append(centroid)

    return np.array(new_centroids)
```

```python
def kmeans(X, k, max_iterations=100):

    indices = np.random.choice(
        len(X),
        k,
        replace=False
    )

    centroids = X[indices].copy()

    for iteration in range(max_iterations):

        clusters = assign_clusters(
            X,
            centroids
        )

        new_centroids = calculate_centroids(
            X,
            clusters,
            k
        )

        if np.allclose(
            centroids,
            new_centroids
        ):
            break

        centroids = new_centroids

    return clusters, centroids
```

```python
clusters, centroids = kmeans(
    X_scaled,
    k=3
)

data["Cluster"] = clusters

print(data)
```
