```python
def euclidean(p1, p2):
    dist_squared = ((p2[0] - p1[0]) ** 2) + ((p2[1] - p1[1]) ** 2)

    return dist_squared ** 0.5
```

```python
def cluster_dist(c1, c2):
    min_dist = float('inf')

    for p1 in c1:
        for p2 in c2:

            dist = euclidean(p1,p2)

            if dist < min_dist:
                min_dist = dist

    return  min_dist
```

```python
def min_dist_clusters(clusters):

    min_dist = float('inf')
    i1 = -1
    i2 = -1

    for i in range(len(clusters)):
        for j in range(i + 1, len(clusters)):

            dist = cluster_dist(clusters[i], clusters[j])

            if dist < min_dist:
                min_dist = dist
                i1 = i
                i2 = j


    return i1, i2, min_dist
```

```python
def merge_clusters(clusters, i1, i2):

    merged_clusters = clusters[i1] + clusters[i2]

    new_clusters = []

    for i in range(len(clusters)):
        if i != i1 and i != i2:
            new_clusters.append(clusters[i])

    new_clusters.append(merged_clusters)

    return new_clusters
```
